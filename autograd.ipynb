```python
import torch

#定义简单的几个变量，相当于神经网络中的权重参数
a=torch.tensor([2.0],requires_grad=True)
b=torch.tensor([-10.0],requires_grad=True)
c=torch.tensor([5.0],requires_grad=True)

#简单模拟一个前向传播
y=a*b+c #2*(-10)+5 = -15

y



```

```python
y.backward() #内置反向传播
```

```python
# 提取梯度
# 我们来看看 PyTorch 算出的 dy/da, dy/db, dy/dc 是多少
print(f"a 的梯度 (dy/da): {a.grad.item()}")  # 期望值应该是 b，即 -10
print(f"b 的梯度 (dy/db): {b.grad.item()}")  # 期望值应该是 a，即 2
print(f"c 的梯度 (dy/dc): {c.grad.item()}")  # 期望值应该是 1   
```
